## two logical questions using operators
### if-else and input

#### 1.movie ticket price calculator

```python
age = int (input("Enter your age:"))
tickets = int(input("Enter number of tickets:"))

if age < 14:
    price = 100
elif age >= 60:
    price = 150
else:
    price = 300

total = price * tickets
print("tickets price:",price)
print("total amount:",total)
    
```

#### 2.electricity usage checker

```python
units = int(input("Enter electricity units"))

if units <= 100:
    print("low usage")
elif units <= 300:
    print("medium usages")
else:
    print("high usage")
```
