## task2
### you have given a number. cheak number is even or odd

```python
num = int(input("Enter a number :"))
if num % 2 == 0:
    print(num,"The number is even")
else:
    print("The number is odd")
```

## task3
### if the number is divisible bt 3 display "fizz" if number is divisible by 5 display "buzz". if number is divisible by 3& 5 both then display "fizz buzz"

```python
num = int(input("Enter a number :"))
if num % 3 == 0 and num % 5 == 0:
    print(num,"FIZZ BUZZ")
elif num % 3 == 0:
    print(num,"FIZZ")
elif num % 5 == 0:
    print(num,"BUZZ")
else:
    print(num,"The number is not divisible by 3 or 5")
```
